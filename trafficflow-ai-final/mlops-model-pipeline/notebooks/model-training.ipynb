```python
# ============================================================
# CHRONOLOGICAL TRAIN / TEST SPLIT
# ============================================================

# Reload original data
df = pd.read_parquet(
    "/kaggle/input/datasets/priyadharshinikaids/mlops-final/ml_data_holiday.parquet"
)

# Convert timestamp
df["timestamp"] = pd.to_datetime(df["timestamp"])

# Sort chronologically
df = df.sort_values("timestamp").reset_index(drop=True)

# Select final features
features = [
    "speedLimit",
    "frc",
    "distance",
    "probeCount",
    "hour",
    "day_of_week",
    "is_weekend",
    "is_peak_hour",
    "avg_temp",
    "wind_speed",
    "air_pressure",
    "rainfall",
    "is_holiday",
    "holiday_type"
]

target = "congestion_level"

# Encode holiday type
df["holiday_type"] = (
    df["holiday_type"]
    .astype("category")
    .cat.codes
)

# Encode target
target_mapping = {
    "Low": 0,
    "Medium": 1,
    "High": 2,
    "Very High": 3
}

df["congestion_level"] = df["congestion_level"].map(target_mapping)

# Create X and y
X = df[features]
y = df[target]

# 80/20 chronological split
split_index = int(len(df) * 0.80)

X_train = X.iloc[:split_index]
X_test = X.iloc[split_index:]

y_train = y.iloc[:split_index]
y_test = y.iloc[split_index:]

print("=" * 70)
print("CHRONOLOGICAL TRAIN / TEST SPLIT")
print("=" * 70)

print("Training:", X_train.shape)
print("Testing :", X_test.shape)

print("\nTraining date range:")
print(df["timestamp"].iloc[0])
print("→")
print(df["timestamp"].iloc[split_index - 1])

print("\nTesting date range:")
print(df["timestamp"].iloc[split_index])
print("→")
print(df["timestamp"].iloc[-1])

print("\nTraining target distribution:")
print(y_train.value_counts().sort_index())

print("\nTesting target distribution:")
print(y_test.value_counts().sort_index())
```

```python
# ============================================================
# CHRONOLOGICAL SPLIT
# ============================================================

n = len(df)

train_end = int(n * 0.70)
val_end = int(n * 0.85)

X_train = X.iloc[:train_end]
y_train = y.iloc[:train_end]

X_val = X.iloc[train_end:val_end]
y_val = y.iloc[train_end:val_end]

X_test = X.iloc[val_end:]
y_test = y.iloc[val_end:]

print("=" * 70)
print("CHRONOLOGICAL DATA SPLIT")
print("=" * 70)

print("Training:", X_train.shape)
print("Validation:", X_val.shape)
print("Testing:", X_test.shape)

print("\nTraining dates:")
print(df["timestamp"].iloc[0], "→", df["timestamp"].iloc[train_end - 1])

print("\nValidation dates:")
print(df["timestamp"].iloc[train_end], "→", df["timestamp"].iloc[val_end - 1])

print("\nTesting dates:")
print(df["timestamp"].iloc[val_end], "→", df["timestamp"].iloc[-1])
```

```python
# ============================================================
# CELL 9B - PREPARE CLASS WEIGHTS FOR TUNED XGBOOST
# ============================================================

import numpy as np

print("=" * 70)
print("PREPARING CLASS WEIGHTS")
print("=" * 70)

# Calculate class counts
class_counts = np.bincount(y_train.astype(int))

total_samples = len(y_train)
num_classes = len(class_counts)

# Balanced class weights
class_weights = {
    cls: total_samples / (num_classes * count)
    for cls, count in enumerate(class_counts)
}

print("\nTraining class counts:")

for cls, count in enumerate(class_counts):
    print(f"Class {cls}: {count:,}")

print("\nBalanced class weights:")

for cls, weight in class_weights.items():
    print(f"Class {cls}: {weight:.4f}")

# Create sample weights
sample_weights = y_train.map(class_weights).to_numpy(dtype=np.float32)

print("\nSample weights created successfully.")

print("Shape:", sample_weights.shape)
print("Data type:", sample_weights.dtype)
print("Minimum weight:", sample_weights.min())
print("Maximum weight:", sample_weights.max())

print("\nReady for tuned XGBoost training.")
```

```python
# ============================================================
# CELL 10 - IMPROVED XGBOOST MODEL WITHOUT CLASS WEIGHTS
# ============================================================

from xgboost import XGBClassifier

print("=" * 70)
print("IMPROVED XGBOOST TRAINING")
print("=" * 70)

improved_xgb = XGBClassifier(
    objective="multi:softprob",
    num_class=4,

    n_estimators=1500,
    learning_rate=0.05,

    max_depth=8,
    min_child_weight=3,

    subsample=0.90,
    colsample_bytree=0.90,

    gamma=0,

    reg_alpha=0.05,
    reg_lambda=1.5,

    tree_method="hist",

    eval_metric="mlogloss",
    early_stopping_rounds=75,

    random_state=42,
    n_jobs=-1
)

print("\nModel configuration:")
print("Estimators       :", 1500)
print("Learning rate    :", 0.05)
print("Max depth        :", 8)
print("Min child weight :", 3)
print("Subsample        :", 0.90)
print("Column sampling  :", 0.90)

print("\nTraining rows   :", len(X_train))
print("Validation rows :", len(X_val))

print("\nStarting training...\n")

# IMPORTANT:
# No sample_weight here
improved_xgb.fit(
    X_train,
    y_train,
    eval_set=[(X_val, y_val)],
    verbose=100
)

print("\n" + "=" * 70)
print("IMPROVED XGBOOST TRAINING COMPLETED")
print("=" * 70)

print("Best iteration:", improved_xgb.best_iteration)
print("Best validation mlogloss:", improved_xgb.best_score)
```

```python
# ============================================================
# CELL 11 - XGBOOST TEST SET EVALUATION
# ============================================================

from sklearn.metrics import (
    accuracy_score,
    f1_score,
    classification_report,
    confusion_matrix
)
import numpy as np

print("=" * 70)
print("XGBOOST TEST SET EVALUATION")
print("=" * 70)

# ------------------------------------------------------------
# 1. Generate predictions
# ------------------------------------------------------------

y_pred = improved_xgb.predict(X_test)

# Convert to integer if necessary
y_pred = np.asarray(y_pred).astype(int)
y_test_eval = np.asarray(y_test).astype(int)

# ------------------------------------------------------------
# 2. Accuracy
# ------------------------------------------------------------

accuracy = accuracy_score(y_test_eval, y_pred)

print("\nTest Accuracy:")
print(f"{accuracy:.4f} ({accuracy * 100:.2f}%)")

# ------------------------------------------------------------
# 3. F1 Scores
# ------------------------------------------------------------

macro_f1 = f1_score(
    y_test_eval,
    y_pred,
    average="macro"
)

weighted_f1 = f1_score(
    y_test_eval,
    y_pred,
    average="weighted"
)

print("\nMacro F1 Score:")
print(f"{macro_f1:.4f}")

print("\nWeighted F1 Score:")
print(f"{weighted_f1:.4f}")

# ------------------------------------------------------------
# 4. Classification Report
# ------------------------------------------------------------

class_names = [
    "Low",
    "Medium",
    "High",
    "Very High"
]

print("\nClassification Report:")

print(
    classification_report(
        y_test_eval,
        y_pred,
        labels=[0, 1, 2, 3],
        target_names=class_names,
        digits=4
    )
)

# ------------------------------------------------------------
# 5. Confusion Matrix
# ------------------------------------------------------------

cm = confusion_matrix(
    y_test_eval,
    y_pred,
    labels=[0, 1, 2, 3]
)

print("\nConfusion Matrix:")
print(cm)

# ------------------------------------------------------------
# 6. Per-class accuracy
# ------------------------------------------------------------

print("\nPer-Class Recall:")

for i, class_name in enumerate(class_names):
    class_total = cm[i].sum()

    if class_total > 0:
        class_recall = cm[i, i] / class_total
    else:
        class_recall = 0

    print(
        f"{class_name:10s}: "
        f"{class_recall:.4f} ({class_recall * 100:.2f}%)"
    )

print("\n" + "=" * 70)
print("EVALUATION COMPLETED")
print("=" * 70)
```
