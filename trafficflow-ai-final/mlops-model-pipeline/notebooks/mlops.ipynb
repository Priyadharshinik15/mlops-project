```python
import pandas as pd

# Keep only columns required for traffic modeling
traffic_hourly = traffic[
    [
        "segmentId",
        "newSegmentId",
        "speedLimit",
        "frc",
        "streetName",
        "distance",
        "segmentProbeCounts"
    ]
].copy()

# Expand the 24 hourly records
traffic_hourly["hourly_data"] = traffic_hourly["segmentProbeCounts"]

traffic_hourly = traffic_hourly.explode("hourly_data", ignore_index=True)

# Extract values from each dictionary
traffic_hourly["timeSet"] = traffic_hourly["hourly_data"].apply(
    lambda x: x["timeSet"]
)

traffic_hourly["dateRange"] = traffic_hourly["hourly_data"].apply(
    lambda x: x["dateRange"]
)

traffic_hourly["probeCount"] = traffic_hourly["hourly_data"].apply(
    lambda x: x["probeCount"]
)

# Remove nested columns
traffic_hourly = traffic_hourly.drop(
    columns=["segmentProbeCounts", "hourly_data"]
)

print("Shape:", traffic_hourly.shape)

print("\nColumns:")
print(traffic_hourly.columns.tolist())

print("\nFirst 10 rows:")
print(traffic_hourly.head(10))
```

```python
import os
import re
import pandas as pd
import geopandas as gpd

traffic_files = sorted([
    f for f in os.listdir(TRAFFIC_PATH)
    if f.endswith(".geojson")
])

all_traffic = []

print("Total traffic files:", len(traffic_files))

for i, filename in enumerate(traffic_files, 1):

    print(f"\nProcessing {i}/{len(traffic_files)}: {filename}")

    filepath = os.path.join(TRAFFIC_PATH, filename)

    # Read GeoJSON
    temp = gpd.read_file(filepath)

    # Remove empty probe-count records
    temp = temp[temp["segmentProbeCounts"].notna()].copy()

    # Keep only required columns
    temp = temp[
        [
            "segmentId",
            "newSegmentId",
            "speedLimit",
            "frc",
            "streetName",
            "distance",
            "segmentProbeCounts"
        ]
    ]

    # Expand 24 hourly records
    temp = temp.explode(
        "segmentProbeCounts",
        ignore_index=True
    )

    # Extract values
    temp["timeSet"] = temp["segmentProbeCounts"].apply(
        lambda x: x["timeSet"]
    )

    temp["dateRange"] = temp["segmentProbeCounts"].apply(
        lambda x: x["dateRange"]
    )

    temp["probeCount"] = temp["segmentProbeCounts"].apply(
        lambda x: x["probeCount"]
    )

    # Remove nested column
    temp = temp.drop(
        columns=["segmentProbeCounts"]
    )

    # Extract date from filename
    date_match = re.search(
        r"(\d{4}-\d{2}-\d{2})",
        filename
    )

    if date_match:
        traffic_date = pd.to_datetime(
            date_match.group(1)
        )
    else:
        print("WARNING: Date not found:", filename)
        continue

    # Convert timeSet → hour
    temp["hour"] = temp["timeSet"] - 2

    # Add date
    temp["date"] = traffic_date

    # Create timestamp
    temp["timestamp"] = (
        temp["date"]
        + pd.to_timedelta(
            temp["hour"],
            unit="h"
        )
    )

    # Keep final columns
    temp = temp[
        [
            "timestamp",
            "date",
            "hour",
            "segmentId",
            "newSegmentId",
            "speedLimit",
            "frc",
            "streetName",
            "distance",
            "probeCount"
        ]
    ]

    print("Rows:", len(temp))

    all_traffic.append(temp)

# Combine all 20 days
traffic_all = pd.concat(
    all_traffic,
    ignore_index=True
)

print("\n==============================")
print("FINAL TRAFFIC DATASET")
print("==============================")

print("Shape:", traffic_all.shape)

print("\nDate range:")
print(traffic_all["date"].min(), "→", traffic_all["date"].max())

print("\nUnique dates:",
      traffic_all["date"].nunique())

print("Unique segments:",
      traffic_all["newSegmentId"].nunique())

print("Unique timestamps:",
      traffic_all["timestamp"].nunique())
```

```python
# Check for potential outliers using the IQR method

numeric_cols = [
    "avg_temp",
    "min_temp",
    "max_temp",
    "wind_speed",
    "air_pressure",
    "elevation",
    "rainfall"
]

outlier_summary = []

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    count = ((df[col] < lower) | (df[col] > upper)).sum()

    outlier_summary.append({
        "Column": col,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower_Bound": lower,
        "Upper_Bound": upper,
        "Outlier_Count": count,
        "Outlier_Percentage": round((count / len(df)) * 100, 2)
    })

outlier_df = pd.DataFrame(outlier_summary)

display(outlier_df)
```
